```python
import random

def display_board(board): # Prints a 3x3 Tic-Tac-Toe Board
    for row in board:
        print(" | ".join(row))
        print("---------")

def player_move(board, player_marker): # Get's the Player's Move
    while True:
        try:
            row = int(input(f"Player {player_marker}, enter row (0, 1, 2): "))
            col = int(input(f"Player {player_marker}, enter column (0, 1, 2): "))
            if 0 <= row <= 2 and 0 <= col <= 2 and board[row][col] == " ":
                return row, col
            else:
                print("Invalid move. Try again.") # Prompts the User for move if invalid
        except ValueError:
            print("Invalid input. Try again.") # Same as above but will not break code if there is a ValueError

def computer_move(board, computer, player): # Makes the Computer's Move
    available_moves = []
    for row in range(3):
        for col in range(3):
            if board[row][col] == " ":
                available_moves.append((row, col))

    # Checks for winning move
    for row, col in available_moves:
        temp_board = [r[:] for r in board] # Creates a temporary copy of the board
        temp_board[row][col] = computer
        if check_win(temp_board, computer):
            return row, col

    # Checks for blocking move
    for row, col in available_moves:
        temp_board = [r[:] for r in board] # Creates a temporary copy of the board
        temp_board[row][col] = player
        if check_win(temp_board, player):
            return row, col

    # If there is no winning or blocking move, chooses a random available move
    return random.choice(available_moves)

def check_win(board, player): # Checks if a Player has won
    # Checks rows
    for row in board:
        if all(cell == player for cell in row):
            return True
    # Checks columns
    for col in range(3):
        if all(board[row][col] == player for row in range(3)):
            return True
    # Checks diagonals
    if all(board[i][i] == player for i in range(3)) or all(board[i][2 - i] == player for i in range(3)):
        return True
    return False

def check_draw(board): # Checks to see if the board is full (Draw)
    return all(board[row][col] != " " for row in range(3) for col in range(3))

def tic_tac_toe(): # Main Function for the Game
    board = [[" " for _ in range(3)] for _ in range(3)] # Initializes an empty 3x3 board

    while True:
        game_mode = input("Play against computer (C) or another player (P)? ").upper() # Allows Player to input desired game mode
        if game_mode in ("C", "P"):
            break
        else:
            print("Invalid choice. Please enter C or P.") # Prompts User for a valid game mode

    while True:
        player_choice = input("Choose X or O: ").upper() # Allows User to pick between playing as "X" or "O"
        if player_choice in ("X", "O"):
            player = player_choice
            computer = "O" if player == "X" else "X"
            break
        else:
            print("Invalid choice. Please enter X or O.") # Prompts User for a Valid Player Option

    game_over = False
    current_turn = player

    print("Welcome to Tic-Tac-Toe!") # Displays a greeting to Player upon start
    display_board(board)

    while not game_over: # Uses a While loop to make sure game is still running in order to Attach Players and Computers Moves to the board
        if current_turn == player:
            if game_mode == "C": # Player vs Computer mode
                row, col = player_move(board, player)
            else: # Player vs Player mode
                row, col = player_move(board, player)
            board[row][col] = player
        else: # Computer's turn
            if game_mode == "C":
                print("Computer's turn...")
                row, col = computer_move(board, computer, player)
                board[row][col] = computer
            else: # Player vs Player mode , Player 2's Turn
                row, col = player_move(board, computer) # Note: 'computer' here acts as the second player's marker
                board[row][col] = computer

        display_board(board)

        if check_win(board, player): # Displays Winner (Player)
            print(f"{player} wins!")
            game_over = True
        elif check_win(board, computer): # Displays Winner (Computer)
            print(f"{computer} wins!")
            game_over = True
        elif check_draw(board): # Displays A Draw
            print("It's a draw!")
            game_over = True

        current_turn = computer if current_turn == player else player # Switches turns

    play_again = input("Play again? (Y/N): ").upper() # Asks Player if they want to play again
    if play_again == "Y":
        tic_tac_toe()

tic_tac_toe()
```
